```python
words = []

with open("English_words.txt", "r") as file:
    for line in file:
        word = line.strip().upper()

        if word.isalpha() and len(word) >= 4:
            words.append(word)

print("Total words loaded:", len(words))
print("Sample words:", words[:10])
```

```python
import random

secret_word = random.choice(words)

print("A secret word has been selected!")
```

```python
stages = [
    """
     -----
     |   |
     O   |
    /|\\  |
    / \\  |
         |
    =========
    """,
    """
     -----
     |   |
     O   |
    /|\\  |
    /    |
         |
    =========
    """,
    """
     -----
     |   |
     O   |
    /|\\  |
         |
         |
    =========
    """,
    """
     -----
     |   |
     O   |
    /|   |
         |
         |
    =========
    """,
    """
     -----
     |   |
     O   |
     |   |
         |
         |
    =========
    """,
    """
     -----
     |   |
     O   |
         |
         |
         |
    =========
    """,
    """
     -----
     |   |
         |
         |
         |
         |
    =========
    """
]

print("Hangman drawings created successfully!")
```

```python
def display_word(secret_word, guessed_letters):

    display = ""

    for letter in secret_word:

        if letter in guessed_letters:
            display = display + letter + " "

        else:
            display = display + "_ "

    return display
```

```python
test_word = "APPLE"
test_guesses = ["A", "P"]

print(display_word(test_word, test_guesses))
```

```python
def play_game():

    # Select a random word
    secret_word = random.choice(words)

    # Store guessed letters
    guessed_letters = []

    # Count wrong attempts
    wrong_attempts = 0

    print("\n===== HANGMAN WORD GUESSING GAME =====")

    while wrong_attempts < 6:

        # Display Hangman drawing
        print(stages[6 - wrong_attempts])

        # Display the hidden word
        print("Word:", display_word(secret_word, guessed_letters))

        print("Guessed letters:", guessed_letters)
        print("Wrong attempts:", wrong_attempts, "/ 6")

        # Get letter from player
        guess = input("Enter a letter: ").strip().upper()

        # Check input
        if len(guess) != 1 or not guess.isalpha():
            print("Please enter only one alphabet letter!")
            continue

        # Check duplicate letter
        if guess in guessed_letters:
            print("You already guessed this letter!")
            continue

        guessed_letters.append(guess)

        # Check whether the guess is correct
        if guess in secret_word:
            print("Correct guess!")
        else:
            wrong_attempts += 1
            print("Wrong guess!")

        # Check winning condition
        if "_" not in display_word(secret_word, guessed_letters):
            print("\nCONGRATULATIONS! YOU WON!")
            print("Correct word:", secret_word)
            return

    # Display losing condition
    print(stages[0])
    print("\nGAME OVER! YOU LOST!")
    print("Correct word:", secret_word)
```

```python
play_game()
```
